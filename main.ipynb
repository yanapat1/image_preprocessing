```python
from utils.color_preprocessing import *
from utils.shape_preprocessing import *
import matplotlib.pyplot as plt
title = ['ResizeAndCrop','ResizeAndCrop + TrimpsCircle','CP','GussianBlurCV','GussianBlurTorch','ScaleRadiusTransform','HSVTransform','SLAHE','PolaLinear']
transform = v2.Compose([
        # ResizeAndCrop(),
        # TrimpsCircle(),
        # CP(),
        # GussianBlurCV() ,
        # GussianBlurTorch(),
        # ScaleRadiusTransform(),
        # HSVTransform(),
        # SLAHE(),
        PolaLinear(),
    ])
img = transform(image).permute(1,2,0).numpy()
lis.append(img)
plt.imshow(img)
```

```python
def show_images_grid(images, title, rows=3, cols=3):
    fig, axes = plt.subplots(rows, cols, figsize=(18, 15))
    for i in range(rows * cols):
        ax = axes[i // cols, i % cols]
        if i < len(images):
            ax.imshow(images[i])
            ax.set_title(str(title[i]), fontsize=22)
            ax.axis('off')
        else:
            ax.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
show_images_grid(lis, title)
```
